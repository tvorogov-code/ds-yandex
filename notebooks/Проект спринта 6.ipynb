- Пропуски в столбцах `eu_sales` и `jp_sales` могут быть случайными, заполняем их средними значениями с учетом платформы и года выпуска:

```python
def approx_sales(row):
    eu_sales = row['eu_sales']
    jp_sales = row['jp_sales']
    
    if pd.isna(eu_sales) or pd.isna(jp_sales):
        group = df[(df['platform'] == row['platform']) & 
                   (df['year_of_release'] == row['year_of_release'])]
        
    if pd.isna(eu_sales):
        eu_sales = group['eu_sales'].mean()
    if pd.isna(jp_sales):
        jp_sales = group['jp_sales'].mean()

    return eu_sales, jp_sales

rv = df.apply(approx_sales, axis=1)
df[['eu_sales', 'jp_sales']] = pd.DataFrame(rv.tolist(), index=rv.index)
```

### 2.4. Обработка пропусков в столбцах `user_score`, `critic_score`, `rating` <a id="2_4"></a>

```python
# Добавление вспомогательных столбцов
df['summary_sales'] = df['na_sales'] + df['eu_sales'] + df['jp_sales'] + df['other_sales'] # Суммарные продажи
df['critic_score_missed'] = df['critic_score'].isna()       # Флаг отсутствия значения в столбце `critic_score`
df['user_score_missed'] = df['user_score'].isna()           # Флаг отсутствия значения в столбце `user_score`
df['jp_sales_category'] = pd.cut(df['jp_sales'], bins=20)   # Категоризация значений в столбце `jp_sales`
# Категоризация по значениям в столбце `summary_sales`
# Квантильная разбивка - в каждой группе примерно поровну строк
df['sum_sales_category'] = pd.qcut(df['summary_sales'], q=int(10), duplicates='drop')
```

- Портативные платформы хуже освещаются на игровых онлайн-порталах, имеют свою закрытую экосистему, в которой долгое время не было пользовательских оценок; также они более распространены в Японии, где культура восприятия игр и оценивания отличается от американской и европейской.
- Игры с низким объемом продаж (малобюджетные или нишевые) могут не получать рецензий со стороны профессиональных критиков. Как следствие, они менее популярны у пользователей, разброс оценок может быть большим, а их количество - недостаточным для формирования итоговой оценки.

Оценка не является следствием случайных наблюдений, а зависит от многих факторов, которые не могут быть формализованы и не сводятся к имеющейся в датасете информации, поэтому правильно будет заполнить пропуски в столбцах `user_score` и `critic_score` индикатором -1:
